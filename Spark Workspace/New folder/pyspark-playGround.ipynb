# **5. how to read dynamic number of columns from text in pyspark**

```python
df1 = spark.read.text('/content/gdrive/MyDrive/spark_workspace/youtube/source/variableData.txt')
df1.show(truncate=False)

df2 = df1.withColumn("splitted",split("value",","))

for i in range(df2.select(max(size("splitted"))).collect()[0][0]):
  df2 = df2.withColumn("col_"+str(i),df2["splitted"][i])

df2.drop("value","splitted").toDF("id","name","place","email","phNo").show()
```

# **19. find difference between two dates**

```python
from pyspark.sql.functions import *

data = [("1","2022-02-25"),("2","2019-06-24"),("3","2019-08-24")]
df=spark.createDataFrame(data=data,schema=["id","date"])

df.select("date",current_date(),datediff(current_date(),col("date"))).show()

df.withColumn("datesDiff", datediff(current_date(),col("date"))) \
  .withColumn("montsDiff", months_between(current_date(),col("date"))) \
  .withColumn("montsDiff_round",round(months_between(current_date(),col("date")),2)) \
  .withColumn("yearsDiff",months_between(current_date(),col("date"))/lit(12)) \
  .withColumn("yearsDiff_round",round(months_between(current_date(),col("date"))/lit(12),2)) \
  .show()
```

# **23. Merge two data frame (scd type1)**

```python
#------------------------------------------------------ STEP-1
schema1 = StructType([
    StructField("sid",IntegerType()),
    StructField("value",IntegerType())
])
data1 = [(1,111),(2,222),(3,333),(4,444),(5,555)]

schema2 = StructType([
    StructField("tid",IntegerType()),
    StructField("value",IntegerType())
])
data2 = [(1,100),(2,200),(6,666),(7,777)]

srcDF = spark.createDataFrame(schema=schema1,data=data1)
tgtDF = spark.createDataFrame(schema=schema2,data=data2)


#------------------------------------------------------ STEP-2
sdf1 = srcDF.withColumn("flag",lit("s"))
tdf1 = tgtDF.withColumn("flag",lit("t"))

mergedf = sdf1.union(tdf1) 
# There is no difference between union and unionall, from spark2.0 it is recommanded to use union
mergedf.show()

#------------------------------------------------------ STEP-3
mergedf1 = mergedf.withColumn("rnk",row_number().over(Window.partitionBy("sid").orderBy("sid","flag")))
mergedf1.filter("rnk == 1").select("sid","value").show()
```

# **25. %calculation to determine special characters in each cell of data frame**

```python
#------------------------------------------------------- STAGE1 (FRAME CREATION)
userSchema = StructType([
    StructField("id",IntegerType()),
    StructField("col1",StringType()),
    StructField("col2",StringType()),
    StructField("col3",StringType())
])
userData = [(1,'A',"@",'X'),(2,'B',"&",'^'),(3,'C',"%",'!')]
df251 = spark.createDataFrame(data=userData, schema=userSchema)

#------------------------------------------------------- STAGE2 (VARIABLE DECLARATION)
res = {}
special_char_count=0
total_rows = df.count()
total_columns = len(df251.columns)
total_cells = total_rows * total_columns

#------------------------------------------------------- STAGE3 (PRINT THE SPECIAL CHARACTER RESULT SET)
for col_name in df251.columns:
    df251= df251.withColumn('matched', when((col(col_name).rlike('[@_!#$%^&*()<>?/|}{~:]')) & ~(col(col_name).rlike('[A-Za-z0-9]')), True).otherwise(False))
    res[col_name] = df251.select('id').where(df251.matched).count()

print(res)

#------------------------------------------------------- STAGE4 (% CALCULATION FOR SPECIAL CHARACTER)
for v in res.values():
  special_char_count+=v

error_percent = (special_char_count/total_cells)*100
print(f'{error_percent} %')
```

# **26. If else condition based on age**

```python
d = {0: "Infant", 1: "Child", 2: "Teenager", 3: "Young", 4:"Adult", 5: "Senior"}

data = [("Alice",25),("Bob",18),("Charlie",68),("Dave",12)]
columns=["name","age"]
df = spark.createDataFrame(data,schema=columns)


df.withColumn("Age_Grouping", when(df.age<13,d[1]) \
                              .when(df.age<20,d[2]) \
                              .when(df.age<30,d[3]) \
                              .otherwise(d[5])).show()
```
